##  Tensors

- In Pytorch, Tensors are the building blocks of neural network.
- A  tensor is a single or multidimensional array of a single data type.
   - A 1-dimensional tensor is often called a vector
   - A 2-dimensional tensor is often called a matrix
   - Tensors with a higher dimension are generally  called a tensor
- Tensors have different types and shapes
- The input, output and parameters of a Neural Network are tensors.
- When processing the input, a neural network  applies a series of tensor operations on the data it receives.

## Tensors Size (or Shape), Number of dimensions and Number of Elements

- Use <code><i>tensor_obj</i>.size()</code> to find out the size of a <code><i>tensor_obj</i></code>.
- Use <code><i>tensor_obj</i>.ndimension()</code> to show its dimension

```python
# One-dimensional tensor object with 5 columns/elements
t = torch.tensor([0, 1, 2, 3, 4])
print("Size: ", t.size())
print("Shape: ", t.shape)
print("Dimension: ", t.ndimension()) # number of dimensions the tensor has
print('Tensor dtype: \t\t', t.dtype)       # type of the data contained within the tensor.
print('Device tensor on: \t', t.device)    # where tensor computations will be performed (CPU or GPU)
print('Tensor layout: \t\t', t.layout)     # how the tensor is stored in memory
print('Number of elements: \t', t.numel()) # number of elements contained within the tensor
```

## Reshaping a tensor
- Use `reshape` or `view`

```python
# One-dimensional tensor object with 5 elements
tensor = torch.tensor([0, 1, 2, 3, 4])
# Two-dimensional tensor object with 5 rows and 1 column.
reshaped_tensor = tensor.view(5, 1)
print("Original Size: ", tensor.size())
print("Size after view method", reshaped_tensor.size())
```

If you do not know the size of your tensor, you can use <b>-1</b> to reshape it (-1 represent any size). Note that you can set only one argument as -1.

### Tensor Addition, Substraction

```python
# Create two sample tensors
u = torch.tensor([1, 0])
v = torch.tensor([0, 1])
```

```python
# Add u and v
u + v
```

```python
# Substract v from u
u-v
```

### Multiplication and Addition with Scalar

```python
# tensor + scalar
u = torch.tensor([1, 2, 3, -1])
v = u + 1
v
```

```python
# tensor * scalar
u = torch.tensor([1, 2])
v = 2 * u
v
```

### Tensor Multiplication
- the two tensors must have the same number of dimensions and the same number of elements

```python
# tensor * tensor

u = torch.tensor([1, 2])
v = torch.tensor([3, 2])
w = u * v
w
```

The result is simply <code>tensor([3, 4])</code>. This result is achieved by multiplying every element in <code>u</code> with the corresponding element in the same position <code>v</code>, which is similar to <i>[1 * 3, 2 * 2]</i>.


### Dot Product

```python
# Calculate dot product of u, v

u = torch.tensor([1, 2])
v = torch.tensor([3, 2])

torch.dot(u,v)
```

The result is <code>tensor(7)</code>. The function is <i>1 x 3 + 2 x 2 = 7</i>.

<h3>Element-wise Product/Hadamard Product</h3>


Multiplication of two tensors corresponds to an element-wise product or Hadamard product.  Consider matrix the <b>X</b> and <b>Y</b> **with the same size**. The Hadamard product corresponds to multiplying each of the elements at the same position, that is, multiplying elements with the same color together. The result is a new matrix that is the same size as matrix <b>X</b> and <b>Y</b> as shown in the following figure:

```python
# Calculate [[1, 0], [0, 1]] * [[2, 1], [1, 2]]

X = torch.tensor([[1, 0], [0, 1]])
Y = torch.tensor([[2, 1], [1, 2]]) 
X_times_Y = X * Y
print("The result of X * Y: ", X_times_Y)
```

<h3>Matrix Multiplication </h3> 

**The number of columns of the matrix on the left side of the multiplication sign must equal to the number of rows of the matrix on the right side.**


First, let us create a tensor <code>X</code> with size 2X3. Then, let us create another tensor <code>Y</code> with size 3X2. Since the number of columns of <code>X</code> is equal to the number of rows of <code>Y</code>. We are able to perform the multiplication.


We use <code>torch.mm()</code> for calculating the multiplication between tensors with different sizes.


```python

A = torch.tensor([[0, 1, 1], [1, 0, 1]])     ## 2 rows, 3 columns
B = torch.tensor([[1, 1], [1, 1], [-1, 1]])  ## 3 rows, 2 columns
A_times_B = torch.mm(A,B)
print("The result of A * B: ", A_times_B)    ## 2 rows, 2 columns
```
