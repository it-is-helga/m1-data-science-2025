```python
with open(filename, "r") as file:
    content = file.read().replace("\n", " ")

with open(filename, "w") as file:
    file.write(content)
```

## Spacy NER

```python
!python -m spacy download en_core_web_lg
```

```python
nlp = spacy.load('en_core_web_lg', disable=["tagger", "parser", "lemmatizer", "attribute_ruler", "tok2vec"])
```

```python
doc_ner = nlp(content)
```

```python
entities_dict = {}

for ent in doc_ner.ents:
    if ent.label_ in {"PERSON", "LOC", "ORG", "GPE", "EVENT", "LANGUAGE", "WORK_OF_ART", "PRODUCT"}:  # Filter relevant entity types
        entity = ent.text
        entity_type = ent.label_

        # Find the sentence containing the entity
        # sentence = ent.sent.text

        # Store in dictionary
        if entity not in entities_dict:
            entities_dict[entity] = (entity_type)

# Save results to CSV
csv_filename = "named_entities.csv"
with open(csv_filename, "w", newline="", encoding="utf-8") as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["Named_Entity", "Type"])  # Header
    
    for entity, entity_type, in entities_dict.items():
        writer.writerow([entity, entity_type,])
```
